## 2. Mixing limit

How much plastic must a pixel hold before it stands out from Guinean ground? For a real plastic spectrum, put `plastic_spectrum.csv` in `4_plastic_content` first, with columns wavelength_nm and reflectance. The USGS Spectral Library has polyethylene. Without it the script uses assumed dips, which are a rough guide only.

```python
!python 4_plastic_content/02_mixing_limit.py
```

**How to read it.** A median site with 20 % plastic gives about 4 % of an EnMAP pixel. If the share needed is above that, the limit explains the result. If it is below, EnMAP should have seen plastic at Guinean sites, so the absence is a finding in itself.

```python
pd.read_csv(ROOT / "4_plastic_content/outputs/02_mixing_limit.csv").round(3)
```
